```python
sns.countplot(x='Booking_Status', data=df)
plt.xticks(rotation=35, ha='right')  # labels ko tilt karega
plt.tight_layout()                   # spacing automatically adjust karega
plt.show()
plt.show()
```

```python
sns.countplot(y='Vehicle_Type', data=df)
plt.title('Booking Status Breakdown')
```

```python
result = df.groupby('Vehicle_Type')['Booking_Value'].sum()
plt.figure(figsize=(15,3))
result.plot(kind='bar')

plt.title('Booking Value by Vehicle Type')

plt.xlabel('Vehicle Type')
plt.ylabel('Total Booking Value')

plt.show()
```

```python
result = df.groupby('Vehicle_Type')['Customer_Rating'].mean()
result.plot(kind='bar')
plt.show()
```

```python
result.plot(kind='pie', autopct='%1.1f%%')
plt.show()
```
